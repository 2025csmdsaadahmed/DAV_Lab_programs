```python
df = pd.read_csv("students.csv")

print("Student Dataset:")
print(df)
```

```python
print("\nFirst 5 rows of the dataset:")
print(df.head())
```

```python
print("\nDataset Information:")
print(df.info())
```

```python
print("\nStatistical Description:")
print(df.describe())
```

```python
print("\n-- Histogram --")

plt.figure(figsize=(8, 5))

plt.hist(df["Marks"], bins=10)

plt.title("Histogram of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python
print("\n-- Box Plot --")

plt.figure(figsize=(8, 5))

sns.boxplot(x=df["Marks"])

plt.title("Box Plot of Student Marks")
plt.xlabel("Marks")

plt.show()
```

```python
print("\n-- Scatter Plot --")

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="StudyHours",
    y="Marks",
    hue="Result"
)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
sns.regplot(data=df,
            x = "StudyHours",
            y="Marks")
palette = {"Pass": "Green", "Fail": "Red"}
scatter=False
plt.show()
```

```python
sns.pairplot(
    df[
        [
            "StudyHours",
            "Attendance",
            "PrevMarks",
            "Marks"
        ]
        ]
)
plt.show()
```

```python
print("\n-- Correlation Matrix --")

correlation = df[
    ["StudyHours", "Attendance", "PrevMarks", "Marks"]
].corr()

# Print the correlation matrix as a table
print(correlation)
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
)

plt.title("Correlation Heat Map")

plt.show()
```
